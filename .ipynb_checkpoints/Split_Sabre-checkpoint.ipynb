```python
'''execute sabre, input: dag: DAG; circ : QuantumCircuit;  coupling_map: CouplingMap. 
  output: cout: number of swaps; initial_layout: the initial mapping that finally determine to use; final_layout_post: the final mapping after execution'''
def execute_sabre(dag, circ, coupling_map, initial_layout):
    seed = np.random.randint(0, np.iinfo(np.int32).max)
    rng = np.random.default_rng(seed)

    physical_qubits = rng.choice(coupling_map.size(), len(dag.qubits), replace=False)
    physical_qubits = rng.permutation(physical_qubits)
    #initial_layout = Layout({q: dag.qubits[i] for i, q in enumerate(physical_qubits)})
    max_iterations = 3
    rev_circ = circ.reverse_ops()
    for i in range(max_iterations):
        for _ in ("forward", "backword"):
            pm = _layout_and_route_passmanager(initial_layout, coupling_map)
            new_circ = pm.run(circ)
        
            pass_final_layout = pm.property_set["final_layout"]
            final_layout = _compose_layouts(initial_layout, pass_final_layout, new_circ.qregs)
            initial_layout = final_layout
            circ, rev_circ = rev_circ, circ

    pm_final = _layout_and_route_passmanager(initial_layout, coupling_map)
    circ_final = pm_final.run(circ)
    cout = circ_final.size() - circ_final.count_ops()['cx']
    pass_final_layout = pm_final.property_set["final_layout"]
    final_layout_post = _compose_layouts(initial_layout, pass_final_layout, circ_final.qregs)
    
    return cout, initial_layout, final_layout_post

'''input: map1_1: dict, the first mapping; map2_1: dict, the second mapping; coupling_graph: nx.Graph; Pqubit: int, number of physical qubit
output: glue_swap: list, contains the swap that need to transform map1_1 to map2_1. '''
                    
def glue_map(map1_1, map2_1, coupling_graph, Pqubit): #swap num
    map1 = copy.deepcopy(map1_1)
    map2 = copy.deepcopy(map2_1)
    for key, value in map1.items():
        if key not in map2 and value not in map2.values():
            map2[key] = value
    notmap = []
    notocc = []
    for i in range(Pqubit):
        if i not in map2:
            notmap.append(i)
        if i not in map2.values():
            notocc.append(i)
    for item in notmap:
        tmp_n = random.choice(notocc)
        map2[item] = tmp_n
        notocc.remove(tmp_n)
 
    reverse_map2 = {v: k for k, v in map2.items()}
    
    graph_first_edge = []
    for item in coupling_graph.edges():
        graph_first_edge.append((reverse_map2[item[0]], reverse_map2[item[1]], 1))
    
    graph_first = rx.PyGraph()
    graph_first.add_nodes_from(range(Pqubit))
    graph_first.add_edges_from(graph_first_edge)
    print("graph new is ", graph_first.edge_list())
    map1_new = dict()
    for key, value in map1.items():
        map1_new[key] = reverse_map2[value]
    print("map1 = ", map1_new)
    token_swapper = ApproximateTokenSwapper(graph_first)
    glue_swap = token_swapper.map(map1_new)
    update_swap = [(map2[item[0]], map2[item[1]]) for item in glue_swap]
    return update_swap
```

```python
def Gauss(A1, b1):
    A = A1.tolist()
    b = b1.tolist()
    n = len(A)
    
    augmented_matrix = [A[i] + [b[i]] for i in range(n)]
    
    for num in range(n):
        k = num
        col = num
        max_r = k
        for i in range(k + 1, n):
            if abs(augmented_matrix[i][col]) > abs(augmented_matrix[max_r][col]):
                max_r=i
        if not max_r ==k:
            for j in range(col, n + 1):
                augmented_matrix[k][j],augmented_matrix[max_r][j] = augmented_matrix[max_r][j], augmented_matrix[k][j]
        if augmented_matrix[k][col]==0:
            k -= 1
            continue
        for i in range(k+1, n):
            if not augmented_matrix[i][col] ==0:
                for j in range(col, n+1):
                    augmented_matrix[i][j]^=augmented_matrix[k][j]
    x = [0] * n
    for i in range(n-1, -1, -1):
        x[i]=augmented_matrix[i][n]
        for j in range(i +1, n):
            x[i] ^= (augmented_matrix[i][j] and x[j])
    
    return x

def rowcol(coupling_graph, M, mapp):

    length = len(M)
    graph = coupling_graph.copy()
    row_handle = list(range(length))
    random.shuffle(row_handle)
    row_has_handle = []
    Cnot_result = []
    rever = {v: k for k, v in mapp.items()}
    
    while len(row_handle) > 1:
        i = row_handle[0]
        row_handle.remove(i)
        row_has_handle.append(i)
        cutpoints = list(nx.articulation_points(graph)) 
        if mapp[i] in cutpoints:
            row_handle.append(i)
            row_has_handle.remove(i)
            continue

        S = []
        S.append(mapp[i])
        for j in row_handle:
            if M[j][i]:
                S.append(mapp[j])
        
        if len(S) >1:
            T = nx.algorithms.approximation.steinertree.steiner_tree(graph, S)
            T_Di = nx.bfs_tree(T, mapp[i])  #Tree, DiGraph, successors or predecessors are iterator
            postorder_result_T = list(nx.dfs_postorder_nodes(T_Di, source=mapp[i]))
            
            for j in postorder_result_T:
                iterator = T_Di.predecessors(j)
                k = next(iterator, None)
                if k is not None:
                    if M[rever[j]][i] == 1 and M[rever[k]][i] == 0:
                        Cnot_result.append([j, k])
                        M[rever[k]] ^= M[rever[j]]

            for j in postorder_result_T:
                iterator = T_Di.successors(j)
                k = next(iterator, None)
                while k is not None:
                    Cnot_result.append([j, k])
                    M[rever[k]] ^= M[rever[j]]
                    k = next(iterator, None)
                    
        subM = np.delete(M, row_has_handle, axis=0)
        subM = np.delete(subM, row_has_handle, axis=1)
        b = M[i, np.delete(np.arange(M.shape[1]), row_has_handle)]
        A = subM.T
        x = Gauss(A, b)
        
        Sprime = []
        sorted_handle = sorted(list(row_handle))
        for j in range(len(x)):
            if x[j] == 1:
                Sprime.append(mapp[sorted_handle[j]])
        
        if len(Sprime) > 0:
            Sprime_root = copy.deepcopy(Sprime)
            Sprime_root.append(mapp[i])
            Tprime = nx.algorithms.approximation.steinertree.steiner_tree(graph, Sprime_root)
            Tprime_Di = nx.bfs_tree(Tprime, mapp[i])
            
            preorder_result_Tprime = list(nx.dfs_preorder_nodes(Tprime_Di, source=mapp[i]))
            for j in preorder_result_Tprime:
                if j not in Sprime:
                    iterator = Tprime_Di.predecessors(j)
                    k = next(iterator, None)
                    if k is not None:
                        Cnot_result.append([j, k])
                        M[rever[k]] ^= M[rever[j]]
            
            postorder_result_Tprime =  list(nx.dfs_postorder_nodes(Tprime_Di, source=mapp[i]))
            for j in postorder_result_Tprime:
                iterator = Tprime_Di.predecessors(j)
                k = next(iterator, None)
                if k is not None:
                    Cnot_result.append([j, k])
                    M[rever[k]] ^= M[rever[j]]

        graph.remove_node(mapp[i])
    
    #print(M)
    return Cnot_result

def glue_map_rowcol(map1_1, map2_1, coupling_graph):  #cnot num
    map1 = copy.deepcopy(map1_1)
    map2 = copy.deepcopy(map2_1)
    #print(map1)
    #print(map2)
    node_num = len(coupling_graph.nodes())
    map_graph_matrix = dict()
    items = 0
    for node in coupling_graph.nodes():
        map_graph_matrix[items] = node
        items += 1

    rever = {v: k for k, v in map_graph_matrix.items()}
    UnOcc_1 = []
    UnOcc_2 = []
      
    for node in coupling_graph.nodes():
        if node not in map1.values():
            UnOcc_1.append(node)
        if node not in map2.values():
            UnOcc_2.append(node)
    
    for key, value in map1.items():
        if key not in map2:
            if value not in map2.values():
                map2[key] = value
                UnOcc_2.remove(value)
            else:
                tmp_n = random.choice(UnOcc_2)
                map2[key] = tmp_n
                UnOcc_2.remove(tmp_n)
    
    for key, value in map2.items():
        if key not in map1:
            if value not in map1.values():
                map1[key] = value
                UnOcc_1.remove(value)
            else:
                tmp_n = random.choice(UnOcc_1)
                map1[key] = tmp_n
                UnOcc_1.remove(tmp_n)
    
    UnOcc_num = len(UnOcc_1)
    for i in range(node_num, node_num + UnOcc_num):
        tmp = UnOcc_1[0]
        map1[i] = tmp
        UnOcc_1.remove(tmp)
    for i in range(node_num, node_num + UnOcc_num):
        tmp = UnOcc_2[0]
        map2[i] = tmp
        UnOcc_2.remove(tmp)
    
    #print(map1)
    #print(map2)
    reverse_map2 = {v: k for k, v in map2.items()}
    perm = [0] * node_num
    for i in range(node_num):
        perm[i] = rever[map1[reverse_map2[map_graph_matrix[i]]]]
    identity_matrix = np.eye(node_num, dtype=int)
    #print("perm", perm)
    permuted_matrix = np.take(identity_matrix, perm, axis=0)
    cnot_result = rowcol(coupling_graph, permuted_matrix,map_graph_matrix)
    
    return cnot_result
    
```

```python
'''directly partition the circuit in average
input: n: int, the number of partitions; C: list, CNOT gates list; coupling_map: CouplingMap; G: nx.Graph, coupling_graph; Pqubit: number of physical qubit
output: data_num: list, contains the number of SWAP for each part and each connection SWAP'''

def partition_many(n, C, coupling_map, G, Pqubit):
    l = len(C)
    part_num = l // n
    C_first = C[:part_num]
    data_num = []
    #max_qubit_index_first = max(max(gate) for gate in C_first)  
    max_qubit_index = max(max(gate) for gate in C) 
    #circ_first =  QuantumCircuit(max_qubit_index_first + 1)
    circ_first = QuantumCircuit(max_qubit_index + 1)
    total = 0
    for cnot in C_first:
        circ_first.cx(cnot[0], cnot[1])
    dag_first = circuit_to_dag(circ_first)
    seed = np.random.randint(0, np.iinfo(np.int32).max)
    rng = np.random.default_rng(seed)
    physical_qubits = rng.choice(coupling_map.size(), len(dag_first.qubits), replace=False)
    physical_qubits = rng.permutation(physical_qubits)
    initial_layout = Layout({q: dag_first.qubits[i] for i, q in enumerate(physical_qubits)})
    cout1, init_first, final_layout_first = execute_sabre(dag_first, circ_first, coupling_map, initial_layout)
    print(len(C_first))
    print("cc", cout1)
    data_num.append(cout1)
    total += cout1
    for i in range(1, n):
        cout2 = 0
        up_swap = 0
        if i == n - 1:
            C_new = C[part_num * i:]
        else:
            C_new = C[part_num * i: part_num * (i+1)]
        #max_qubit_index_new = max(max(gate) for gate in C_new) 
        #circ_new = QuantumCircuit(max_qubit_index_new + 1)
        circ_new = QuantumCircuit(max_qubit_index + 1)
        for cnot in C_new:
            circ_new.cx(cnot[0], cnot[1])
        dag_new = circuit_to_dag(circ_new)
        mappre = layout2map(final_layout_first)
        cout2, init_sec, final_layout_sec = execute_sabre(dag_new, circ_new, coupling_map, final_layout_first)  
        mappost = layout2map(init_sec)
        minswap = 1000000
        mincnot = 1000000
        for times in range(100):
            up_swap = glue_map(mappre, mappost, G, Pqubit)
            #print("update_swap", len(up_swap))
            if len(up_swap) < minswap:
                minswap = len(up_swap)
            up_cnot = glue_map_rowcol(mappre, mappost, G)
            if len(up_cnot) < mincnot:
                mincnot = len(up_cnot)
        print("update_swap", minswap)
        print("compare to cnot", 3 *minswap)
        print("update_cnot", mincnot)
        total += minswap
        data_num.append(minswap)
        data_num.append(3 * minswap)
        data_num.append(mincnot)
        
        print("cc", cout2)
        data_num.append(cout2)
        total += cout2        
        final_layout_first = final_layout_sec
    print("total: ", total)
    data_num.append(total)
    return data_num
```

```python
def compose_mapping(map1, map2): #would like to return map1 \circ map2^{-1} that is, mappost, mappre
    map2_inverse = {v: k for k, v in map2.items()}
    compose_map = dict()
    for k, v in map2_inverse.items():
        if v in map1:
            compose_map[k] = map1[v]
    return compose_map

def connect_mapping(mappre, mappost, graph):
    UnOcc_post = []
    for node in graph.nodes():
        if node not in mappost.values():
            UnOcc_post.append(node)
            
    for key, value in mappre.items():
        if key not in mappost:
            if value not in mappost.values():
                mappost[key] = value
                UnOcc_post.remove(value)
            else:
                shortest_paths = nx.single_source_dijkstra(graph, value)
                nearest_point = min((node for node in UnOcc_post), key=lambda node: shortest_paths[1][node])
                mappost[key] = nearest_point
                UnOcc_post.remove(nearest_point)
        
    perm = compose_mapping(mappost, mappre)
    pygraph = rx.networkx_converter(graph)
    token_swapper = ApproximateTokenSwapper(pygraph)
    swapping = token_swapper.map(perm)
    
    return swapping

map1 = {'A': 0, 'B' : 1, 'C': 2}
map2 = {'D': 1, 'B' : 2, 'C' : 3}
print(connect_mapping(map1, map2,graph))
```
